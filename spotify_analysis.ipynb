# Spotify Songs — Exploratory Data Analysis
Dataset: 114,000 Spotify tracks  
Tools: Python, Pandas, Matplotlib  
Goal: Explore what makes a song popular on Spotify

## Loading the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("spotify.csv")
df.head()
```

## Understanding the Data

```python
print(df.columns)
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.shape)
```

## Checking missing values

```python
df.isnull().sum()
```

## Removing missing values

```python
df.dropna(inplace=True)
```

## Removing Duplicate Values

```python
df = df.drop_duplicates(subset='track_id')
```

## Which Songs Are Most Popular?

```python
df.sort_values('popularity',ascending=False).head(10)
```

## How is song popularity distributed across Spotify tracks?

```python
df['popularity'].describe()
```

```python
plt.figure(figsize=(15,8))
plt.hist(df['popularity'],bins=20,edgecolor='black',linewidth=1)
plt.title('Popularity')
plt.xlabel("Popularity")
plt.ylabel("Frequency")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()
```

```python
(df['popularity']==0).sum()
```

### Insight

Most songs have low to moderate popularity scores, while only a small number of tracks achieve very high popularity. The presence of songs with zero popularity indicates that many tracks receive little to no listener engagement.

## Does Musical Key of a song influence its Popularity?

```python
df.groupby('key')['popularity'].mean()
```

### Insight

Some keys show slightly higher average popularity than others, but the differences are relatively small. Musical key alone does not appear to be a strong factor influencing song popularity.

## Which Artists have the highest average song Popularity?

```python
df.groupby('artists')['popularity'].mean().sort_values(ascending=False).head(10)
```

```python
artists_popularity=df.groupby('artists')['popularity'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(15,8))
plt.bar(artists_popularity.index,artists_popularity.values,color= 'red' ,edgecolor='black',linewidth=1)
plt.title('Artists Popularity')
plt.xlabel("Artists")
plt.ylabel("Popularity")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()
```

### Insight

A small group of artists consistently achieves high popularity, suggesting that artist recognition and fan following can have a significant impact on song success.

## Which music genres tend to produce more popular songs?

```python
genre_popularity=df.groupby('track_genre')['popularity'].mean().sort_values(ascending=False)
top_genre=genre_popularity.head(10)
print(top_genre)
```

```python
plt.figure(figsize=(15,8))
plt.bar(top_genre.index,top_genre.values,color= 'red' ,edgecolor='black',linewidth=1)
plt.title('Average popularity of top genre')
plt.xlabel("Genre")
plt.ylabel("Average Popularity")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()
```

### Insight

Popularity varies across genres, indicating that audience preferences differ by genre. Some genres consistently attract higher listener engagement than others.

## Do explicit songs perform better than non-explicit songs?

```python
df.groupby('explicit')['popularity'].mean()
```

```python
explicit=df.groupby('explicit')['popularity'].mean()

plt.figure(figsize=(15,8))
plt.bar(['non explicit','explicit'],explicit.values,color= 'red' ,edgecolor='black',linewidth=1)
plt.title('Average Popularity: Explicit vs Non-Explicit Songs')
plt.xlabel("Song Type")
plt.ylabel("Popularity")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()
```

### Insight

The average popularity of explicit and non-explicit songs differs, suggesting that content type may have some influence on listener engagement. However, the difference is relatively small, indicating that explicit content alone is unlikely to have a major impact on song popularity.

## Does the musical mode of a song affect its popularity?

```python
df.groupby('mode')['popularity'].mean()
```

```python
mode_popularity=df.groupby('mode')['popularity'].mean()

plt.figure(figsize=(15,8))
plt.bar(['minor','major'],mode_popularity.values,color= 'red' ,edgecolor='black',linewidth=1)
plt.title('Average Popularity of major and minor mode')
plt.xlabel("mode")
plt.ylabel("Average Popularity")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()
```

### Insight

Songs in major and minor modes show similar popularity levels, indicating that musical mode is not a strong factor affecting popularity.

## Which audio features are most closely related to song popularity?

A correlation matrix is used to examine relationships between popularity and Spotify audio features.

```python
numerical_cols=['popularity','duration_ms','danceability','energy','loudness','speechiness','acousticness','instrumentalness','liveness','valence','tempo']
df[numerical_cols].corr()['popularity'].sort_values(ascending=False)
```

### Insight

Loudness (0.07) and danceability (0.06) have the highest positive correlations, suggesting that louder and more danceable songs tend to be slightly more popular. On the other hand, instrumentalness (-0.13) has the strongest negative correlation, indicating that songs with fewer vocals are generally less popular. However, since all correlation values are close to zero, these relationships are weak and audio features alone cannot reliably explain a song's popularity.

## Does song duration influence popularity?

```python
duration_min=df['duration_ms']/60000
duration_min.describe()
```

```python
plt.figure(figsize=(15,8))
plt.scatter(duration_min,df['popularity'],alpha=0.3)
plt.title('Duration v/s Popularity')
plt.xlabel("Duration")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

### Insight

Most songs in the dataset are tightly packed between ~2.9 to 4.4 minutes, and this is also where most of the popular songs are found. There is no clear direct relationship between duration and popularity, but very long songs are rare and usually not highly popular

## Are energetic songs more popular?

```python
plt.figure(figsize=(15,8))
plt.scatter(df['energy'],df['popularity'],alpha=0.3)
plt.title('Energy v/s Popularity')
plt.xlabel("Energy")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

### Insight

There is no strong direct relationship between energy and popularity. Songs across all energy levels (low to high) show a wide spread of popularity values.However, we can notice a slight tendency where moderately to high-energy songs appear more frequently in higher popularity ranges

## Are more danceable songs generally more popular?

```python
plt.figure(figsize=(15,8))
plt.scatter(df['danceability'],df['popularity'],alpha=0.3)
plt.title('Danceability v/s Popularity')
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

### Insight

Many highly popular songs have moderate to high danceability values, suggesting a positive relationship. However, danceability alone does not fully explain song popularity.

## Final Conclusion & Key Insights

From the analysis of Spotify tracks, it is observed that more popular songs tend to share certain tendencies rather than strict rules. Songs with moderate duration (around 2–5 minutes), higher danceability, and good energy levels are more frequently associated with higher popularity scores. Additionally, tracks in the minor mode show a slightly higher tendency to be more popular compared to those in the major mode. However, no single audio feature strongly determines popularity, indicating that listener engagement is influenced by a combination of musical attributes along with external factors such as trends, artists, and exposure. Therefore, while clear patterns exist, there is no fixed formula that guarantees a hit song on Spotify.
